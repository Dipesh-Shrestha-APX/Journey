# LEARNING THE REGEX

```python
import re
```

```python
text = '''
          $4.3 billion is the net worth to be achieved
          but the
          thing is that FY 1032 WAS $23.555 billion is the most this guy can take
          but the amount fy2022 IS THE KING OF of $123 billion
       '''

# SOME BASICS
    # [ABC] it means any element inside the bracket is selected
    # [ABC]+ it means any combination of them is to be shift_selection_mode
    # [ABC]* same as above but will select the empty space as well
    # () it means after applying the regex select the content inside it and return
    # \d it means numbers
    # \d{10} it means numbers repeated 10 times {} represents repetitions
    # [^abc] it means except the a,b,c elements
    # [1-9] it means numbers 1 to 9
    # (?i)apple it means case insensitive extraction of word apple
    # [\s]* any combination of space i.e white space or nothing
       # \s it means white space
       # \S it means anything except white space

pattern = '(?i)fy[\s]*[\d]{4}[^\$]+\$[\d\.]+ bILLion'

  #[^\$]+ means after that fyDATE any text except $ upto $ in any combination is to
  #be extracted
```

```python
# this findall funciton of re can has flags like IGNORECASE to disregard the CASE SENSITIVITY
amount_mentioned = re.findall(pattern, text, flags=re.IGNORECASE) # Here flags one works for the billion
first_selection = re.search(pattern, text, flags=re.IGNORECASE)
print(amount_mentioned)
print(first_selection.group()) # search help to see only first selection
```
