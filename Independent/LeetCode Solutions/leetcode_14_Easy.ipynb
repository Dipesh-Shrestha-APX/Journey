```python
def longestCommonPrefix(strs):
    smallestStr = strs[0]
    if len(strs) ==1:
        return strs[0]
    for i in range(len(strs)):
        if len(strs[i]) < len(smallestStr):
            smallestStr = strs[i]
    index = []
    less = 1
    for str in (strs):
        for j in range(len(smallestStr)):
            if smallestStr[j] != str[j]:
                less = 0
                print(smallestStr[j])
                print(str[j])
                print(j)
                index.append(j)
                break
    if len(index) == 0:
        index = [0]
    smallestIndex = min(index)
    if less == 1:
        smallestIndex = len(smallestStr)
    return smallestStr[0:smallestIndex]
```

```python
ans = longestCommonPrefix(["flower","flow","flight"])
ans2 = longestCommonPrefix(["abca","aba","aaab"])
ans3 = longestCommonPrefix(["reflower","flow","flight"])
ans4 = longestCommonPrefix(["", ""])
```
