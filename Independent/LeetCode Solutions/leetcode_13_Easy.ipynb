# *ROMAN NUMERALS TO VALUE CONVERSION*

```python
def romanToInt(s):
    diction = {
            'I':1,
            'V':5,
            'X':10,
            'L':50,
            'C':100,
            'D':500,
            'M':1000
    }

    # To generate a list collection
    # i.e. group of numeral letter e.g. IV X like that separately
    # from the roman numeral strings
    list =[]
    two_flag = 0
    for i in range(len(s)):
      if i<(len(s)-1):
        # print(s[i])
        # print(diction[s[i]])
        # print(diction[s[i+1]])

        ## THIS HANDLES THE REST
        if (diction[s[i]]<diction[s[i+1]]):
          list.append(s[i:i+2])
          two_flag = 1
        # elif (diction[s[i]]==diction[s[i+1]]):
        #   list.append(s[i])
        #   two_flag = 1
        else:
          if two_flag == 0:
            list.append(s[i])
          two_flag = 0

    ## If we ended with small Large numeral like IV nothing is needed to be don
    # since both are going to be added to the list
    # but for the equal like II or Large small like VI last numeral is left
    # hence adding it to the list
    if two_flag == 0:
      list.append(s[-1])
    print(list)

    # Now assigning the right values to those numeral and taking a sum

    sumValue = 0
    for i in range(len(list)):
      if(len(list[i])==3):
        sumValue += 3*diction[list[i][0]]
      elif(len(list[i]) ==2):
        sumValue += -1*diction[list[i][0]]
        sumValue += diction[list[i][1]]
      else:
        sumValue += diction[list[i]]
    # print(sumValue)

    return sumValue
```

```python
value = romanToInt("II")
print(value)
```
