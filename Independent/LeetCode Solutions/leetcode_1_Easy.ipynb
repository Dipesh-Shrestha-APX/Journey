# *INDICES OF THE NUMBER THAT ADDS UPTO TARGET IN A GIVEN LIST*

```python
def twoSum(self, nums, target):
  for i in range(len(nums)):
      for j in range(i, len(nums)):
          if i!=j:
              if (nums[i]+nums[j] == target):
                  ans = [i,j]
                  return ans

  return [-1]
```
