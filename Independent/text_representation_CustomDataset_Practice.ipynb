# *DATA IMPORTS*

```python
import pandas as pd
```

```python
dfNew = pd.read_csv('/content/drive/MyDrive/WebScrapedCSVs/movies.csv')
```

```python
df = pd.DataFrame({})
df['overview'] = dfNew['overview'].iloc[:100]
df['overview'] = df['overview'].fillna('')
df['overview'].head()
```

# *OHE*

```python
# OHE requires categorical variable
```

```python
# 100 categorical values from ChatGPT
data = [
    "apple","orange","banana","apple","banana","orange","orange","apple","banana","orange",
    "apple","apple","banana","orange","apple","banana","orange","banana","apple","orange",
    "banana","apple","orange","orange","banana","apple","orange","banana","orange","apple",
    "banana","apple","orange","banana","apple","orange","apple","banana","orange","banana",
    "apple","orange","apple","banana","orange","apple","banana","orange","apple","apple",
    "banana","orange","banana","apple","orange","apple","banana","orange","banana","apple",
    "apple","orange","banana","apple","orange","banana","apple","orange","banana","orange",
    "apple","banana","apple","orange","banana","apple","orange","orange","banana","apple",
    "banana","orange","apple","orange","apple","banana","orange","apple","banana","apple",
    "orange","banana","apple","orange","banana","orange","apple","banana","apple","orange"
]
```

```python
df['fruits'] = data
df.head()
```

```python
from sklearn.preprocessing import OneHotEncoder
oheObj = OneHotEncoder()
ohevectors = oheObj.fit_transform(df[['fruits']])
dfFruits = pd.DataFrame(
    ohevectors.toarray(),
    columns = oheObj.get_feature_names_out()
)
dfFinal = pd.concat([df, dfFruits], axis=1)
dfFinal
```

# *BAG OF WORDS*

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cvObj = CountVectorizer(max_features=100)
      # THE max_features means the top 100 words
      # with most frequency from the vocabulary
      # are to be used for creating bow vectors
bow = cvObj.fit_transform(df['overview'])
```

```python
print((bow[0]).shape)
bow[0].toarray()
```

```python
len(cvObj.vocabulary_)
```

# *N-Gram*

```python
## BIGRAM
cvObj2 = CountVectorizer(ngram_range=(2,2), max_features=100)
bigrams = cvObj2.fit_transform(df['overview'])
cvObj2.vocabulary_
```

```python
# BOTH UNIGRAM AND BIGRAM
cvObj2 = CountVectorizer(ngram_range=(1,2), max_features=100)
bigrams = cvObj2.fit_transform(df['overview'])
cvObj2.vocabulary_
```

# *TF-IDF*

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidfObj = TfidfVectorizer(max_features =100)
tfidfVectors = tfidfObj.fit_transform(df['overview'])
print(tfidfVectors.toarray())
```

# *Final Practice*

```python
# OHE
df['fruits']
from sklearn.preprocessing import OneHotEncoder
oheObj = OneHotEncoder()
oheVectors = oheObj.fit_transform(df[['fruits']])
oheVectors.toarray()
dfnew = pd.DataFrame(
    oheVectors.toarray(),
    columns = oheObj.get_feature_names_out()
)
dffinal = pd.concat([df, dfnew], axis=1)
dffinal

# ngram
from sklearn.feature_extraction.text import CountVectorizer
cvObj = CountVectorizer(ngram_range=(1,3), max_features =5)
cvVectors = cvObj.fit_transform(df['overview'])
cvVectors.toarray()
dffinal['1-3-Grams'] = str(cvVectors.toarray())
dfnew = pd.DataFrame(
    cvVectors.toarray(),
    columns = cvObj.get_feature_names_out()
)
dffinal2 = pd.concat([dffinal, dfnew],axis=1)
dffinal2

# tfidf
from sklearn.feature_extraction.text import TfidfVectorizer
tfidfObj = TfidfVectorizer(max_features=3)
tfidfVectors = tfidfObj.fit_transform(df['overview'])
dffinal2['tfidf'] = str(tfidfVectors.toarray())
dffinal2
dfnew = pd.DataFrame(
    tfidfVectors.toarray(),
    columns = tfidfObj.get_feature_names_out()
)
tfidfObj.get_feature_names_out().shape
dffinal3 = pd.concat([dffinal2, dfnew], axis=1)
dffinal3
```
