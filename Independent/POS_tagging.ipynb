```python
import spacy
```

```python
# Load the dictionary and form an dictionary object of spacy
actualDictObj = spacy.load('en_core_web_sm')
```

```python
# Use that dictionary as reference and any text you pass it will be processed and
# it will keep the features provided by spacy saved after make some internal evaluation
# which could be accessed using the functions and such
# i.e. perform the operation on the text provided with the dictionary reference and
# save that to the new Object
doc = actualDictObj("Apple is the tastiest fruit or is it ?")
```

```python
# Now you can access the features of it for an example
for word in doc:
  coarsePOS = word.pos_
  finePOS = word.tag_
  detailedPOS = spacy.explain(finePOS)
  print(f"{word} ------ Coarse grained POS: {coarsePOS} ------ fine grained POS: {finePOS} ------ detailed POS: {detailedPOS}")
```

```python
# this displacy can help you visualize how the POS tagging was done
from spacy import displacy
```

```python
displacy.render(doc, style='dep', jupyter = True)
```

```python
# to ease the visualization
# you can use the options
# setting the options
myOptions = {'distance': 150}
```

```python
displacy.render(doc, style='dep', options= myOptions)
```
