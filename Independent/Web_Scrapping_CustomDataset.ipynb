```python
contentDictList =[]
for i in range(len(contents)):
  # print('++++++++++++')
  # print(contents[i])
  # print('++++++++++++')
  bsObj = BeautifulSoup(contents[i], 'lxml') # lxml is the parser
  # print(bsObj.prettify())
  # print(f"at {i} value = {type(bsObj.find('p'))}")
  # print (str(type(bsObj.find('p'))) != "<class 'NoneType'>")
  if (str(type(bsObj.find('p'))) != "<class 'NoneType'>"):
    json_text = bsObj.find('p').get_text(strip= True)
    contentDict = json.loads(json_text)
    contentDictList.append(contentDict)

adult = []
genre_ids = []
original_language = []
original_title = []
overview = []
popularity = []
release_date = []
title = []
vote_average = []
vote_count = []
for content in contentDictList:
  for item in content['results']:
    adult.append(item['adult'])
    genre_ids.append(item['genre_ids'])
    original_language.append(item['original_language'])
    original_title.append(item['original_title'])
    overview.append(item['overview'])
    popularity.append(item['popularity'])
    release_date.append(item['release_date'])
    title.append(item['title'])
    vote_average.append(item['vote_average'])
    vote_count.append(item['vote_count'])

d = {'adult':adult, 'genre_ids':genre_ids, 'original_language':original_language, 'overview':overview, 'popularity':popularity, 'release_date':release_date, 'title':title, 'vote_average':vote_average, 'vote_count':vote_count}
df = pd.DataFrame(d)
len(df)
df.head()
```
