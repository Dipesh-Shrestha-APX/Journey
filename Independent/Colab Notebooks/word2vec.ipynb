# *Creating a WORD2VEC model through CustomDataset*

```python
import kagglehub
# Download latest version
path = kagglehub.dataset_download("rupanshukapoor/harry-potter-books")
print("Path to dataset files:", path)
```

```python
import os
folder = (os.listdir(path))
print(folder)
```

```python
path2 = os.path.join(path,folder[0])
print(path2)
path3 = os.listdir(path2)
print(path3)
```

```python
path4 = os.path.join(path2,path3[0])
print(path4)
print(os.listdir(path4))
```

```python
finalpath = path4
```

```python
from nltk import sent_tokenize
import nltk
nltk.download('punkt_tab')
from gensim.utils import simple_preprocess
```

```python
output = []
i = 1
for filename in (os.listdir(path4)):
  # if i == 1:
  f = open(os.path.join(path4,filename))
  corpus = f.read()
  raw_sent = sent_tokenize(corpus)
  # print(raw_sent)
  for sent in raw_sent:
    # print(sent)
    output.append(simple_preprocess(sent))
    # i = 0
  # break
print(output)
```

```python
len(output.find('dumbledoer'))
```

```python
model = Word2Vec(
    window= 50,
    min_count = 5
)
model.build_vocab(output)
```

```python
model.train(output, total_examples = model.corpus_count, epochs = model.epochs)
```

```python
print(model.wv['potter'])
print(model.wv.similarity('snape','dumbledore'))
print(model.wv.most_similar('dumbledore'))
```
