# *EMOJI REMOVAL*

```python
## THE PATTERN SECTION IS ONLY COPIED FROM CHATGPT NOTHING ELSE
## SINCE IT'S A STANDARD
def emojiRemove(text):
  emoji_pattern = re.compile(
      "["
      "\U0001F600-\U0001F64F"  # Emoticons
      "\U0001F300-\U0001F5FF"  # Symbols & pictographs
      "\U0001F680-\U0001F6FF"  # Transport & map symbols
      "\U0001F1E0-\U0001F1FF"  # Flags
      "\U00002702-\U000027B0"  # Dingbats
      "\U000024C2-\U0001F251"  # Enclosed characters
      "]+",
      flags=re.UNICODE)
  return emoji_pattern.sub('',text)
```

```python
def tokenizeWordLevel(text):
  tokens = dictionary(text)
  out = []
  for token in tokens:
    out.append(str(token))
  return out
def tokenizeSentenceLevel(text):
  tokens = tokenizeWordLevel(text)
  # print(tokens)
  sentence = []
  out_list = []
  for token in tokens:
    # print(type(token))
    # token = str(token)
        # since we will get string's list from the tokenizeWordLevel
        # no need to convert for comparision
    if token == '.':
      out_list.append(" ".join(sentence))
      sentence = []
    else:
      sentence.append(token)
  return out_list
```

# *Stemming and Lemmatization*

```python
# !pip install nltk
import nltk
```

```python
# STEMMING
from nltk.stem.porter import PorterStemmer
psObj = PorterStemmer()
def stemmer(text):
  tokens = tokenizeWordLevel(text)
  output = []
  for token in tokens:
    output.append(psObj.stem(token))
  return " ".join(output)
```

```python
# LEMMATIZATION
  # Note needs to be in lower case to work properly
from nltk.stem import WordNetLemmatizer
wnlObj = WordNetLemmatizer()
def posLabelForLemmatize(pos):
  if pos.startswith('J'): # adjective
    return 'a'
  elif pos.startswith('V'): # verb
    return 'v'
  elif pos.startswith('N'): # noun
    return 'n'
  elif pos.startswith('R'): # adverb
    return 'r'
  else:
    return 'n'
def lemmatizer(text):
  tokens = tokenizeWordLevel(text)
  # print(tokens)
  token_pos_list = nltk.pos_tag(tokens)
  output = []
  for token, pos in token_pos_list:
    # print(pos)
    output.append(wnlObj.lemmatize(token, pos= posLabelForLemmatize(pos)))
  return " ".join(output)
```
