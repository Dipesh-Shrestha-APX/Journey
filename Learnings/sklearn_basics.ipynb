# Sometimes you may want to perform different transformation in different columns


   ___### ColumnTransformer



```python
import pandas as pd

# Create sample data
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Age": [25, 30, 22, 28, 35],
    "City": ["Kathmandu", "Pokhara", "Biratnagar", "Lalitpur", "Bhaktapur"],
    "Score": [88.5, 92.0, 76.5, 85.0, 90.3],
    "Passed": [True, True, False, True, True]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display it
df
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.preprocessing import OneHotEncoder
column_trans = ColumnTransformer(
    [
        ('City_Category', OneHotEncoder(dtype='int'),['City']),
        ('Name_Vectors', CountVectorizer(), 'Name')
    ],
    remainder='passthrough'
)
```

```python
column_trans.fit(df)
```

```python
column_trans.get_feature_names_out()
```

```python
column_trans.transform(df)
```

## TRANSFOMERS AND ESTIMATORS can be combined together into a single unifying object called Pipeline

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
# CREATION OF PIPELINE OBJECT
pipe = make_pipeline(
    StandardScaler(), # All the rest should be transformer
    LogisticRegression() # Last one can be estimator/transformer
)
```

```python
X, y = load_iris(return_X_y = True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
pipe.fit(X_train,y_train)
```

```python
# WE CAN USE IT LIKE ANY OTHER ESTIMATOR
accuracy_score(pipe.predict(X_test), y_test)
```

# AUTOMATED PARAMETER SEARCH
## i.e. Hyperparameter Tuning

```python
from sklearn.datasets import fetch_california_housing
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV
from sklearn.model_selection import train_test_split
from scipy.stats import randint
```

```python
X, y = fetch_california_housing(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=0)

#Define the parameter space that will be seaerched over
# defining the parameter boundary for the hyperparameter space
param_distributions = {
    'n_estimators':randint(1,5),
    'max_depth':randint(5,10)
}

#Create a search object
search = RandomizedSearchCV(
    estimator=RandomForestRegressor(random_state=0),
    n_iter=5,
    param_distributions=param_distributions,
    random_state=0
)


#Perform the search
search.fit(X_train, y_train)
```

```python
# e.g
# Iteration	n_estimators	max_depth	Meaning
# 1	        3	            7	          Build 3 trees, each with depth ≤ 7
# 2	        2	            9	          Build 2 trees, each with depth ≤ 9
# 3	        4	            6	          Build 4 trees, each with depth ≤ 6
# 4	        1	            5	          Build 1 tree, with depth ≤ 5
# 5	        2	            8	          Build 2 trees, each with depth ≤ 8
```

```python
search.best_params_
```

```python
search.score(X_test,y_test)
#It's corresponding to the best params
```
