```python
# Labels tile in the plot graph
plt.plot(batsman['index'], batsman['V Kohli'], linestyle = 'dashed', linewidth = 10)
plt.plot(batsman['index'], batsman['RG Sharma'], color= "red", linestyle="dashdot", marker = 'D', markersize=5)

plt.title("Rohit Sharma vs Virat Kohli runs in different years")
plt.xlabel('SEASON')
plt.ylabel('Runs')
```

LEGENDS

```python
# Naming the lines
plt.plot(batsman['index'], batsman['V Kohli'], linewidth = 5,label="V Kohli")
plt.plot(batsman['index'], batsman['RG Sharma'], linewidth = 5,label='RG Sharma')

plt.title("Rohit Sharma vs Virat Kohli runs in different years")
plt.xlabel('SEASON')
plt.ylabel('Runs')
plt.legend()
plt.grid()
```

BAR PLOT

```python
# BIVARIATE
#Categorical VS NUMERICAL(AGGREGATE OPERATION)
#USE CASE: AGGREGATE ANALYSIS
```

```python
#SIMPLE BARCHART

colors = ["Blue", "Pink", "Green", "Yellow", "Red", "Purple", "Orange", "Violet", "Black", "White"]
children_count = [5, 8, 6, 4, 7, 3, 2, 1, 9, 10]

plt.bar(colors, children_count, color = "RED")
#Way of dealing with larger names of the category is simply  use
# roatation from ticks
plt.xticks(rotation=90)

#print only the graph
plt.show()
```

```python
#Another of dealing with the no. of categories is high and name is also longer
plt.barh(colors, children_count, color = "GREEN")
```

```python
# COLOR AND LABEL

df = pd.read_csv("/content/batsman_season_record.csv")
df
```

```python
np.arange(df.shape[0])
```

```python
#Side by side BAR Chart plt
#Multiple BAR CHART

# at x axis plot the with no. from no. of rows
# and then shift the center of that bar 0.2 from the original center
# and have it's width 0.2 so that it would extend 0.1 on left and right
# which would allow another 0.2 width bar to fit in
# draw according to df['season_year'] values on y axis

plt.bar(np.arange(df.shape[0]) -0.2, df['2015'], width=0.2)
plt.bar(np.arange(df.shape[0]), df['2016'], width=0.2)
plt.bar(np.arange(df.shape[0]) +0.2, df['2017'], width=0.2)

#Repalce the x-axis labels with the correseponding batsman names
plt.xticks(np.arange(df.shape[0]), df["batsman"])

#It allows to showcase only the graphs
plt.show()
```

STACKED BAR CHART

```python
df
```

```python
plt.bar(df["batsman"], df["2015"], color = "red", label="2015")

#Bottom should start from the height of ----
plt.bar(df["batsman"], df["2016"], bottom = df["2015"], color = "blue", label ="2016")
plt.bar(df["batsman"], df["2017"], bottom = df["2016"]+df["2015"], color = "green",label ="2017")
plt.legend()
plt.show()
```
