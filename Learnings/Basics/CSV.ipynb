 Opening A CSV file from an URL

```python
import requests
from io import StringIO

url = "https://raw.githubusercontent.com/paiv/fci-breeds/main/fci-breeds.csv"

# If the site requires headers (sometimes needed for GitHub raw), you can add them
headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "text/csv"
}

response = requests.get(url, headers=headers)

# Now treat the content as CSV text
csv_data = StringIO(response.text)
df2 = pd.read_csv(csv_data)

print(df2.head(5))
```

CONVERTORS

```python
def rename(name):
  if name == "Apple":
    return "APPL"
  else:
    return name

pd.read_csv("/content/Fortune500_Cleaned.csv",converters={'Company Name':rename}).head()
```
