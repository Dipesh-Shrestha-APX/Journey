```python
from sklearn.datasets import make_classification
import torch
```

```python
# Creating a synthetic classification dataset using sklearn
X,y =make_classification(
    n_samples=10, #No of samples
    n_features=2, #No of features
    n_informative=2, #No of informative features
    n_redundant=0, #No of redundant features
    n_classes=2, #No of classes
    random_state=42 # For reproducibility
)
```

```python
X.shape
```

```python
X = torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.long)
```

```python
from torch.utils.data import Dataset, DataLoader
```

```python
class CustomDataset(Dataset):
  def __init__(self, features, labels):
    self.features = features
    self.labels = labels
  def __len__(self):
    return len(self.features)
  def __getitem__(self, idx):
    #Transformation is performed here

    #
    return self.features[idx], self.labels[idx]
```

```python
dataset = CustomDataset(X, y)
```

```python
len(dataset)
```

```python
dataset[0]
```

```python
#DataLoader is an iterable
dataloader = DataLoader(dataset, batch_size=2, shuffle=True)
```

```python
for batch_features, batch_labels in dataloader:
  print(batch_features, batch_labels)
  print("--------------------------")
```
