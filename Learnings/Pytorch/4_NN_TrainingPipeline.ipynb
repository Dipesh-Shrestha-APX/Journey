```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
```

```python
df.head()
```

```python
df.shape
```

```python
df.drop(columns=['id','Unnamed: 32'], inplace= True)
```

```python
df.shape
```

```python
df.head()
```

```python
#Train Test split
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,1:],df.iloc[:,0],test_size=0.2,random_state=42)
```

```python
#Scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_test
```

```python
#Label encoding the y
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
y_train
```

```python
import torch
```

```python
# Numpy arrays to tensors
X_train_tensor = torch.from_numpy(X_train.astype(np.float32))
X_test_tensor = torch.from_numpy(X_test.astype(np.float32))
y_train_tensor = torch.from_numpy(y_train.astype(np.float32))
y_test_tensor = torch.from_numpy(y_test.astype(np.float32))
```

```python
y_train_tensor
```

```python
#Defining the model
########### From video4
# class MySimpleNet():
#   def __init__(self,X):
#     self.weights = torch.rand(X.shape[1], 1, dtype= torch.float64, requires_grad= True)
#     self.bias = torch.zeros(1, dtype= torch.float64, requires_grad= True)
#   #Forward pass
#   def forward(self,X):
#     z = torch.matmul(X, self.weights) + self.bias
#     y_pred = torch.sigmoid(z)
#     return y_pred
#   #loss calculation by BinaryCrossEntropy
#   def loss_Function(self,y_pred,y):
#     loss = -(y_train_tensor*torch.log(y_pred) + (1-y_train_tensor)*torch.log(1-y_pred)).mean()
#     return loss

########## From video5 updation using the nn.Module
import torch.nn as nn
class MySimpleNet(nn.Module):
  def __init__(self, num_features):
    super().__init__()
    self.linear = nn.Linear(num_features, 1)
    self.sigmoid = nn.Sigmoid()
  #Forward pass
  def forward(self,features):
    out = self.linear(features)
    out = self.sigmoid(out)
    return out
  #loss calculation by BinaryCrossEntropy
  # def loss_Function(self,y_pred,y):
  #   loss = -(y_train_tensor*torch.log(y_pred) + (1-y_train_tensor)*torch.log(1-y_pred)).mean()
  #   return loss
  ##Using a builtin loss functin of BCE from nn.Module
```

```python
# Defining the Parameters for the SGD ||1 and ||2
learning_rate = 0.1
epochs = 30
```

```python
#Defining the loss function
loss_function = nn.BCELoss() ##Using built-in loss function
```

```python
# Create a model
model = MySimpleNet(X_train_tensor.shape[1])
# model.weights.shape
# model.bias.shape

#Defining the optimizer
optimizer = torch.optim.SGD(model.parameters(), lr= learning_rate)
      # model.parameters() is an iterator that will access the weights
      # and bias

#Define loop
for epoch in range(epochs):
  #Forward pass
  y_pred = model(X_train_tensor)

  ##LossCalc 1
  # loss = model.loss_Function(y_pred,y_train_tensor)

  ##Using inbuilt loss function from nn.Module 2
  loss = loss_function(y_pred,y_train_tensor.view(-1,1))
      # the y_train_tensor was 1D whereas y_pred was scalar so reshaping it

  ## Making gradient zero gradient cal and weight updation || 2
  optimizer.zero_grad()

  #Backward pass
  loss.backward()

  ##zero gradients || 1
  # model.linear.weight.grad.zero_()
  # model.linear.bias.grad.zero_()
  ##Weights update 1
  # with torch.no_grad():
  #   model.linear.weight -= learning_rate * model.linear.weight.grad
  #   model.linear.bias -= learning_rate * model.linear.bias.grad

  ## Making gradient zero gradient cal and weight updation || 2 cont..
  optimizer.step()

  #Print loss in each epoch
  print(f"Epoch={epoch+1}, Loss={loss.item()}")
```

```python
model.linear.weight
model.linear.bias
```

```python
#Model evaluation
with torch.no_grad():
  y_pred = model(X_test_tensor)
  y_pred= (y_pred > 0.5).float() #i.e. >0.5 = 1 else 0
  accuracy = (y_pred == y_test_tensor).float().mean()
  print(f"Accuracy={accuracy.item()}")
# y_pred
# y_test_tensor
```
