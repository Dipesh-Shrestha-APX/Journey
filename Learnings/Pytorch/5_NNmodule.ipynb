```python
# create dataset
features = torch.rand(10,5)
# create model
model = Model(features.shape[1])
#call model for forward pass
  # by __call__ will automatically call forward function
  # through below syntax ## a magic function inside the nn.Module
model(features) #Standard way of calling forward method
#option was model.forward(features)
```
