Eg_2 :

for y = x^2

for z = sin(y)

finding the differential using Autograd


```python
x = torch.tensor(3.0, requires_grad=True)
print(x)
y = x**2
print(y)
z = torch.sin(y)
print(z)
```

```python
#ask for the y's gradient to be retained as well
#during future gradient calculation
y.retain_grad()
#calculate the gradients
z.backward()
#print the gradients of z wrt x i.e dz/dx
print(x.grad)
```

```python
#This is the leaf differential gradient retained
#that's the gradient of z wrt y i.e. dz/dy
y.grad
```

```python
#PROOF
y = torch.tensor(9.0, requires_grad=True)
print(y)
z = torch.sin(y)
print(z)
z.backward()
y.grad
```

AUTO_GRAD FINAL EXAMPLE

```python
import torch
```

```python
#Inputs
x = torch.tensor(6.7)
y = torch.tensor(0.0)

w = torch.tensor(1.0)
b = torch.tensor(0.0)
```

```python
#Binart CrossEntropy Loss calculation code
def binary_cross_entropy_loss(prediction, target):
  epsilon = 1e-8
  prediction = torch.clamp(prediction, epsilon, 1-epsilon)
  return -target*torch.log(prediction) - (1-target)*torch.log(1-prediction)
```

```python
#Forward Pass
z = w*x+b
y_pred = torch.sigmoid(z)
loss = binary_cross_entropy_loss(y_pred, y)
```

```python
#Derivatives for backpropagation
dloss_dy_pred = (y_pred-y)/(y_pred*(1-y_pred))
dy_pred_dz = y_pred*(1-y_pred)
dz_dw = x
dz_db = 1
dL_dw = dloss_dy_pred*dy_pred_dz*dz_dw
dL_db = dloss_dy_pred*dy_pred_dz*dz_db
```

```python
print(f"Manual Gradient of loxs wrt w: {dL_dw}")
print(f"Manual Gradient of loxs wrt b: {dL_db}")
```

DISABLE GRADIENT TRACKING

```python
# option 1 - requires_grad(False)
x = torch.tensor(2.0, requires_grad=True)
y = x**2
print(x)
x.requires_grad_(False)
print(x)
```

```python
# option 2 - detach()
x = torch.tensor(2.0, requires_grad=True)
z = x.detach()
y = x**2
print(y)
y = z**2
print(y)
```

```python
# option /3 - torch.no_grad()
x = torch.tensor(2.0, requires_grad=True)
y = x**2
print(y)
with torch.no_grad():
  y = x**2
print(y)
```
